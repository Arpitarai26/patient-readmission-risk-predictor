```python
import pandas as pd

from src.preprocess import (
    replace_question_marks,
    remove_leakage_columns
)

from src.feature_engineering import *
```

```python
DATA_PATH = "../data/diabetic_data.csv"

df = pd.read_csv(DATA_PATH)

df = replace_question_marks(df)

df = remove_leakage_columns(df)

print(df.shape)
```

```python
df = create_total_visits(df)

df = create_total_meds_ratio(df)

df = create_prior_visits_flag(df)

df = create_high_glucose_flag(df)

df = create_polypharmacy_flag(df)

df = create_meds_per_day(df)

df = create_long_stay_flag(df)

df = create_chronic_patient_flag(df)

df = create_frequent_visitor_flag(df)

df = create_utilization_score(df)

df = create_clinical_complexity_score(df)

df = create_interaction_features(df)
```

```python
new_columns = [
    col
    for col in df.columns
    if col not in [
        "readmitted"
    ]
]

print(
    f"Total columns: {len(df.columns)}"
)
```

```python
engineered_features = [
    "total_visits",
    "total_meds_ratio",
    "prior_visits_flag",
    "high_glucose_flag",
    "polypharmacy",
    "meds_per_day",
    "long_stay_flag",
    "chronic_patient_flag",
    "frequent_visitor_flag",
    "utilization_score",
    "clinical_complexity_score",
    "stay_medication_interaction",
    "diagnosis_medication_interaction",
    "inpatient_emergency_interaction"
]

df[engineered_features].head()
```

```python
feature_summary = pd.DataFrame({
    "feature": engineered_features,
    "dtype": [
        str(df[col].dtype)
        for col in engineered_features
    ]
})

feature_summary
```

```python
feature_summary.to_csv(
    "../reports/feature_summary.csv",
    index=False
)

print("Feature summary saved.")
```
